```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('student_performance.csv')
df.head()
```

```python
# here we are checking the data type of student id column
df['Student_ID'].dtypes
```

```python
# as the dtype of gender column is string, if you have observed the Gender column it has smaller case values
# so we are converting them uppercase using str.upper() method
df['Gender'].str.upper()
```

```python
# herewe are converting them to lowercase using str.lower() method
df['Gender'].str.lower()
```

```python
# strip the prefix
# her we are replacing the 'S' in the student_id column with A
df['Student_ID'].str.replace('S', 'A')
```

```python
# boolean mask, like isin() but for substrings

# here we are checking if the string '1' is present in the Student_ID column
df['Student_ID'].str.contains('1')  
```

```python
data = {'Name': ['Peter Parker', 'Baby Johnson', 'Santoor Mummy'],
       'Age': [25, 30, 35],
       'City': ['New York', 'London', 'Paris']}
df = pd.DataFrame(data)
df
```

```python
# here we are splitting the Name column into two columns, First Name and Last Name
# as our previous dataset studentperformance.csv does not have a name column, we are creating a new dataset with name column
df['Name'].str.split(' ', expand=True)          
```

```python
# here we are removing the white space from the Name column using str.strip() method
# as our our Name column does not have any white space, you are not seeing any changes in the output, but if you have a white space in the Name column, it will be removed

df['Name'].str.strip()    # remove whitespace
```
