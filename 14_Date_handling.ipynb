# Date Handling

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('countryLockdowndates.csv')
df.head(5)
```

```python
# these are the columsn which we have in our dataset
df.columns.tolist()
```

```python
# checking the data type of the Date column
df['Date'].dtype
```

```python
# as in the previous step we saw that the datatype of Date column is string, 
# so we need to convert it to datetime format so that we can perform date operations on it.
df['Date'] = pd.to_datetime(df['Date'])   # convert string column to real datetime
```

```python
# here we are extracting Year from Date column and creating a new column called Year in our dataframe
df['Year'] = df['Date'].dt.year
df.head()
```

```python
# here we are extracting month from Date column and creating a new column called Month in our dataframe
df['Month'] = df['Date'].dt.month
df.head()
```

```python
# here we are finding the name of the day by using date column
# for that we will be using .dt.day_name() function which will return the weekday name for that date.
df['Weekday'] = df['Date'].dt.day_name()
df.head()
```

```python
# here it give only the short name of the weekday like Mon, Tue, Wed, Thu, Fri, Sat, Sun
df['Weekday'] = df['Date'].dt.strftime('%a')
```

```python
# as you have seen in our previous step, the weekday column is in string format, 
# so we will create a new column called Weekday_Num which will have the weekday number 
# for that we will be using .dt.dayofweek function which will return the weekday number for that date.
"""
where Monday = 0, Tuesday = 1, Wednesday = 2, Thursday = 3, Friday = 4, Saturday = 5, Sunday = 6
"""
df['Weekday_Num'] = df['Date'].dt.dayofweek
df.head()
```

```python
# here we are filtering the data by using date range
df[(df['Date'] >= '2020-01-01') & (df['Date'] <= '2020-06-30')]
```
